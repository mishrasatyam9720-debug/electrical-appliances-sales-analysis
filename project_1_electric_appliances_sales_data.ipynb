```
import pandas as pd
df = pd.read_csv("sales_data.csv")
df
```

```
df["Total_sales"]=df['Units_Sold']*df['Unit_Price']
df
```

```
def calculate_discount(x):
  if x>100000:
    return x*0.1
  else:
    return 0
df["discount"]=df["Total_sales"].apply(calculate_discount)
df
```

```
df["total_cost"]=df["Cost_Price"]*df["Units_Sold"]
df
```

```
df["total_profit"]=df["Total_sales"]-df["total_cost"]-df["discount"]
df
```

```
def profit_status(y):
  if y>0:
    return "profit"
  else:
    return "loss"
df["profit_status"]=df["total_profit"].apply(profit_status)
df
```

```
def stock_status(row):
  x = row['Units_Sold']
  y = row['total_profit']
  if x>=10 and y>0:
    return "keep in stock"
  else:
    return "don't keep in stock"
df["stock_status"]=df.apply(stock_status, axis=1)
df
```

```
df["Net_sales"] = df["Total_sales"] - df["discount"]
df
```

```
top = df.loc[df["Net_sales"].idxmax()]
print(top)
```

```
top = df.loc[df["Net_sales"].idxmax()]
print("===== SALES SUMMARY =====\n")

print("TOP PRODUCT :", top["Product"])
print("NET SALES :", top["Net_sales"])
print("PROFIT :", top["total_profit"])
print("STATUS :", top["profit_status"])

print("\nKEEP IN STOCK:")
print(df[df["stock_status"] == "keep in stock"]["Product"].to_list())

print("\nDON'T KEEP IN STOCK:")
print(df[df["stock_status"] == "don't keep in stock"]["Product"].to_list())
```
